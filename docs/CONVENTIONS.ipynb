# Conventions

These rules are what makes the code leak-free. Read once, then keep them open while you code.

## Timing

- Row `t` **predicts the return `r_t`**.
- Every feature at row `t` is computed from data **up to and including t-1**. In pandas that means a `shift(1)` or more on anything that contains `r_t` or later information.
- Target for horizon `h`: `y_t = r_t + r_{t+1} + ... + r_{t+h-1}` (the return earned over the next `h` periods starting at row `t`). Its label interval is `[t, t+h-1]`; `label_end_times` returns `t1 = date of the last return included`.
- For `h = 1`, `y_t = r_t` and `t1 = t`.

## Data shapes

- Single asset: `pd.Series` or `pd.DataFrame` with a sorted `DatetimeIndex`.
- Cross-section: `pd.DataFrame` with a `MultiIndex` named `["date", "asset"]`, sorted by date then asset.
- Splitters expose `split(X)` and yield `(train_idx, test_idx)` as integer position arrays.
- Models expose `fit(X, y) -> self` and `predict(X) -> np.ndarray`. Fitted attributes end with an underscore (`coef_`, `intercept_`).

## Evaluation

- **OOS R-squared:** `1 - sum((y - yhat)^2) / sum((y - b)^2)` where `b` is the benchmark forecast. Gu-Kelly-Xiu use `b = 0` for individual stock returns; for aggregate series people often use the prevailing (expanding) mean. Pick one, state it in the README, and keep it fixed.
- **IC:** the per-date Spearman correlation between prediction and realized return, then averaged (and report its t-stat).
- **Sharpe:** annualize explicitly (`sqrt(252)` for daily, `sqrt(12)` for monthly); always report net of costs when a portfolio is traded.

## Never

- Never shuffle time-series data before splitting.
- Never fit scalers, PCA, imputers or feature selection on rows outside the training fold.
- Never choose hyperparameters on the test window.
- Never report a model without a baseline and a trial count.
